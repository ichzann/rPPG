```python
def normalize_matrix(matrix):
    for i in range(3):
        channel = matrix[:, :, i] 
        
        mean = np.mean(channel)
        std = np.std(channel)
        
        matrix[:, :, i] = (channel - mean) / (std + 1e-8)
    return matrix
```
